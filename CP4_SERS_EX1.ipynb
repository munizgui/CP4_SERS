Uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

```python
# 2. Renomeie Appliances para Consumo_Eletrodomesticos
dados.rename(columns={'Appliances':'Consumo_Eletrodomesticos',
                    'T1':'Temperatura_1',
                    'T2':'Temperatura_2',
                    'T3':'Temperatura_3'
                    }, inplace=True)
```

```python
# 3. Determine o maior consumo de eletrodomésticos registrado na amostra.
dados['Consumo_Eletrodomesticos'].max()
```

```python
print(f'O maior consumo de eletrodomésticos registrado na amostra foi {dados['Consumo_Eletrodomesticos'].max()} Wh')
```

```python
# 4. Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo somente os registros acima desse limite.

# Cálculo do valor máximo e limiar utilizando o nome atualizado da coluna
valor_maximo = dados["Consumo_Eletrodomesticos"].max()
limiar = valor_maximo * 0.70

# Filtragem do DataFrame
df_acima_limite = dados[dados["Consumo_Eletrodomesticos"] > limiar]

print("Valor máximo:", valor_maximo)
print("Limiar:", limiar)
print("Quantidade de registros:", len(df_acima_limite))

df_acima_limite
```

```python
# 5. Conte os registros selecionados e calcule qual percentual da amostra eles representam.

# parte 1. Total de registros na amostra original
total_registros = len(dados)

# parte 2. Contagem dos registros acima do limiar
registros_selecionados = len(df_acima_limite)

# parte3. Cálculo do percentual
percentual = (registros_selecionados / total_registros) * 100

# parte 4. Exibição dos resultados
print(f"Total de registros na amostra: {total_registros}")
print(f"Registros acima do limiar: {registros_selecionados}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

```python
# 6. Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.

# parte 1. Média da temperatura T1 (Temperatura_1)
media_t1 = dados["Temperatura_1"].mean()

# parte 2. Limiar de 70% do consumo máximo
limiar_consumo = dados["Consumo_Eletrodomesticos"].max() * 0.70

# parte 3. Filtragem com ambas as condições simultâneas
df_consumo_e_temp = dados[
    (dados["Consumo_Eletrodomesticos"] > limiar_consumo) &
    (dados["Temperatura_1"] > media_t1)
]

print(f"Temperatura média T1: {media_t1:.2f} °C")
print(f"Registros encontrados: {len(df_consumo_e_temp)}")

df_consumo_e_temp
```
