# Dataset 2 — Steel Industry Energy Consumption (UCI)

Fonte: https://archive.ics.uci.edu/dataset/851/steel+industry+energy+consumption

Situação

As informações coletadas são da DAEWOO Steel Co. Ltd., em Gwangyang, Coreia do Sul. A empresa produz diversos tipos de bobinas, chapas de aço e chapas de ferro. Os dados sobre consumo de energia elétrica estão armazenados em um sistema baseado em nuvem. As informações sobre o consumo de energia da indústria são armazenadas no site da Korea Electric Power Corporation (pccs.kepco.go.kr), e as perspectivas dos dados diários, mensais e anuais são calculadas e apresentadas.

```python
import pandas as pd
```

```python
# 1. Carregue a amostra e renomeie Usage_kWh para Consumo_kWh, simplificando também os nomes dos atributos de fator de potência.


df = pd.read_csv('Steel_industry_data_SAMPLE02.csv')

df = df.rename(columns={
    'Usage_kWh': 'Consumo_kWh',
    'Lagging_Current_Power_Factor': 'FP_Atrasado',
    'Leading_Current_Power_Factor': 'FP_Adiantado'
})

print("Colunas atuais do DataFrame:")
print(df.columns.tolist())
```

```python
# 3. Determine o maior consumo registrado e calcule um limiar de 75% do consumo máximo.

maior_consumo = df['Consumo_kWh'].max()
limiar_75 = maior_consumo * 0.75

print(f"Maior consumo registrado: {maior_consumo:.2f} kWh")
print(f"Limiar de 75% do consumo máximo: {limiar_75:.2f} kWh")
```

```python
# 4. Crie um DataFrame contendo os registros acima desse limiar e calcule a quantidade e o percentual de registros.

df_alto_consumo = df[df['Consumo_kWh'] > limiar_75]

qtd_registros_alto = len(df_alto_consumo)
percentual_alto = (qtd_registros_alto / len(df)) * 100

print(f"Quantidade de registros acima do limiar: {qtd_registros_alto}")
print(f"Percentual em relação ao total: {percentual_alto:.2f}%")
```

```python
# 5. Verifique quantos registros desse DataFrame pertencem à categoria Maximum Load.

qtd_max_load = len(df_alto_consumo[df_alto_consumo['Load_Type'] == 'Maximum_Load'])

print(f"Registros de alto consumo que pertencem à categoria Maximum Load: {qtd_max_load}")
```

```python
# 6. Escolha um dos atributos de fator de potência, observe seus valores e defina um limite coerente para representar valores mais baixos

print("Distribuição dos valores de FP_Atrasado no grupo de alto consumo:")
display(df_alto_consumo['FP_Atrasado'].describe())


limite_fp = 90.0

print(f"\nFator de Potência escolhido: FP_Atrasado")
print(f"Limite definido para representar valores mais baixos: {limite_fp}")
```

```python
# 7. Crie um novo DataFrame contendo simultaneamente consumo elevado e fator de potência abaixo do limite escolhido.

df_critico = df_alto_consumo[df_alto_consumo['FP_Atrasado'] < limite_fp]

print(f"Quantidade de registros com consumo elevado E fator de potência baixo: {len(df_critico)}")
display(df_critico.head())
```

# 8. Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.

###A combinação de consumo elevado (pico de demanda) com um fator de potência baixo é a pior situação para uma instalação elétrica industrial.Risco Financeiro: Como os dados de consumo de energia da indústria são monitorados pela concessionária Korea Electric Power Corporation, a empresa corre grande risco de pagar multas severas por exceder o limite de energia reativa justamente nos momentos em que mais puxa energia da rede.  Sobrecarga: Um fator de potência baixo exige que correntes maiores circulem pelos cabos e transformadores para realizar o mesmo trabalho útil na produção das chapas de aço. Como o consumo (kWh) já está no seu limite máximo (acima de 75%), essa corrente extra gera superaquecimento excessivo, podendo causar falhas e paradas na produção da fábrica. Identificar o que está ligado nesses momentos exatos ajuda a equipe a instalar bancos de capacitores adequados.
