Dataset 3 - Power Consumption of Tetouan City (UCI)

fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

Situação

Consumo elétrico de três zonas de distribuição da cidade de Tétouan, associado a variáveis meteorológicas.

```python
import pandas as pd
```

```python
df3 = pd.read_csv("Tetuan City power consumption_SAMPLE03 1.csv")
```

1-RENOMEANDO AS ZONAS

```python
df3 = df3.rename(columns={
    "Zone 1 Power Consumption": "Consumo_Zona_1",
    "Zone 2  Power Consumption": "Consumo_Zona_2",
    "Zone 3  Power Consumption": "Consumo_Zona_3"
})
```

```python
df3.columns
```

2-MAIORES VALORES DE CADA ZONA

```python
maximos = df3[
    ["Consumo_Zona_1", "Consumo_Zona_2", "Consumo_Zona_3"]
].max()
print(f"os maiores valores de cada zona são, respectivamente: {maximos}")
```

3-ZONA COM MAIOR PICO DE CONSUMO

```python
zona_maior_pico = maximos.idxmax()
maior_pico = maximos.max()

print(f"Zona com maior pico: {zona_maior_pico}")
print(f"Valor máximo: {maior_pico}")
```

4-70% DO VALOR MAXIMO

```python
limiar = 0.70 * maior_pico

print(f"Limiar (70% do máximo): {limiar}")
```

```python
df3_acima_limiar = df3[df3[zona_maior_pico] > limiar].copy()
print(df3_acima_limiar)
```

5-CONTAR OS REGISTROS E CALCULAR OS PERCENTUAIS

```python
quantidade = len(df3_acima_limiar)
total = len(df3)
percentual = (quantidade / total) * 100

print(f"\nRegistros acima do limiar: {quantidade}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

6-TEMPERATURA MÉDIA

```python
temperatura_media = df3["Temperature"].mean()

print(f"Temperatura média: {temperatura_media:.2f}")
```

```python
df3_consumo_temperatura = df3[
    (df3[zona_maior_pico] > limiar) &
    (df3["Temperature"] > temperatura_media)
].copy()

quantidade_com_temperatura = len(df3_consumo_temperatura)
percentual_com_temperatura = (quantidade_com_temperatura / total) * 100

print(f"\nRegistros acima do limiar e acima da temperatura média: "
      f"{quantidade_com_temperatura}")
print(f"Percentual da amostra: {percentual_com_temperatura:.2f}%")
```

7- COMPARAÇÂO DOS CONJUNTOS


```python
print("\nComparação: Na interpretação final, se a quantidade de registros diminuir após adicionar a temperatura, isso significa que nem todos os períodos de consumo elevado ocorreram sob temperaturas acima da média. A condição ambiental funciona, portanto, como um filtro adicional e torna o conjunto analisado mais restritivo.\n")
print(f"Somente consumo acima do limiar: {quantidade} registros")
print(f"Consumo + temperatura acima da média: "
      f"{quantidade_com_temperatura} registros")
print(f"Redução: {quantidade - quantidade_com_temperatura} registros")
```
