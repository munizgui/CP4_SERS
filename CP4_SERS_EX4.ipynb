Situação

Dados de geração e sensores de duas usinas fotovoltaicas, com medições realizadas em inversores e sensores ambientais.

```python
import pandas as pd
```

```python
df4 = pd.read_csv("Plant_2_Generation_Data_SAMPLE04(in).csv")
```

```python
df4 = df4.rename(columns={
    "DC_POWER": "Potencia_CC",
    "AC_POWER": "Potencia_AC",
    "DAILY_YIELD": "Geracao_Diaria"
})
```

MAIOR POTÊNCIA CA

```python
potencia_ca_max = df4["Potencia_AC"].max()
print(f"Maior Potencia_CA registrada: {potencia_ca_max}")
```

CALCULO DO LIMIAR CORRESPONDENTE A 70%


```python
limiar_alta_geracao = potencia_ca_max * 0.70
print(f"Limiar (70%): {limiar_alta_geracao:.2f}")
```

CRIANDO DATAFRAME E CALCULE O PERCENTUAL

```python
df_alta_geracao = df4[df4['Potencia_AC'] > limiar_alta_geracao]

qtd_alta = len(df_alta_geracao)
pct_alta = (qtd_alta / len(df4)) * 100

print(f"Quantidade de registros de alta geração: {qtd_alta}")
print(f"Percentual em relação ao total: {pct_alta:.2f}%")
```

INDENTIFICANDO FREQUÊNCIA DE INVERSOR

```python
frequencia_inversores = df_alta_geracao['SOURCE_KEY'].value_counts()

print("--- Frequência de Inversores em Alta Geração ---")
print(frequencia_inversores)
print("O resultado permite observar que este inversor específico atinge picos de geração máxima com maior frequência ou sustenta a operação em alta potência por mais tempo do que os outros dentro desta amostra.")
```
